```
# this self test was suggested by Claude, thanks Claude!
print("--- valid expressions ---")
examples = [
  "exp(-theta**2)",
  "sin(x)**2 + cos(x)**2",
  "2 * pi * r",
  "sqrt(x**2 + y**2)",
  "1 / (1 + exp(-x))",
  "DiracDelta(x - 1)",
  "Heaviside(t)",
]
for ex in examples:
  print(f"{ex!r:30} -> {parse.sympyExpression(ex)}")

print("\n--- rejected / unsafe expressions ---")
malicious = [
  "__import__('os').system('echo pwned')",
  "().__class__.__bases__[0].__subclasses__()",
  "open('/etc/passwd').read()",
  "[x for x in range(10)]",
  "lambda: 1",
  "os.system('echo pwned')",
]
for ex in malicious:
  try:
    parse.sympyExpression(ex)
    raise RuntimeError(f"{ex!r:45} -> unsafe expression NOT rejected")
  except parse.UnsafeExpressionError as e:
    print(f"{ex!r:45} -> rejected: {e}")
```

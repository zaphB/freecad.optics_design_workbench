```
from numpy import *
from matplotlib.pyplot import *
import yaml
```

```
with open('benchmark-scores.yaml', 'r') as _f:
  scores = yaml.safe_load(_f)
```

```
figure(figsize=(14,8))
ufile = sorted(list(set([f for run in scores for _,_,f,_,_,_ in run]))) 
allT = [mean([r[0] for r in run]) for run in scores]
for run in scores:
  for t, verInfo, file, binary, workerCount, raysPerHourPerWorkerLog in run:
    T, R = array(raysPerHourPerWorkerLog).T
    plot((T-min(T))/60, R,
         marker='x+*^D'[ufile.index(file)],
         ls='-' if binary.lower()=='freecad' else ':',
         color=cm.jet((t-min(allT))/(max(allT)-min(allT))),
         lw=workerCount, alpha=min([1, 2/workerCount ]) )
xlabel('minutes since simulation start')
ylabel('rays/s per worker')
semilogy()
text(.5, 1.02, 'line width: worker count, markers: '+', '.join([f'{'x+*^D'[i]} : {f}' for i,f in enumerate(ufile)]),
     transform=gca().transAxes, ha='center')
c = colorbar(cm.ScalarMappable(None, cm.jet), cmap=cm.jet, ax=gca(), shrink=.5, aspect=39)
c.set_label('time that benchmark was run')
tickPos = [allT[argmin(abs(t-allT))] for t in linspace(min(allT), max(allT), 5)]
c.set_ticks((tickPos-min(tickPos))/(max(tickPos)-min(tickPos)))
c.set_ticklabels([time.strftime('%m.%Y', time.localtime(t)) for t in tickPos])
tight_layout()
```
